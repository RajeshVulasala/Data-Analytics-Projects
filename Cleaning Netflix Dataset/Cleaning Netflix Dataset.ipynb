```python
# missing value columns
missing = df.isnull().sum()
missing[missing > 0]
```

```python
# insted of removing the categorical columns replace them with unknown
# because the director, cast, country, rating are very importent in movies daatadest
categorical_cols = df[["director", "cast", "country", "rating"]]
categorical_cols.isnull().sum()
```

```python
# replacing the missing or null data with unknown
cols_to_fill = ["director", "cast", "country", "rating"]
for col in cols_to_fill:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# as you can see wen we used describe the there is only one col with numerical data so we have one more numerical data if we observe the data carefully that is 
# "date_added" it is in object datatype so we have to parse it into datetime data
df["date_added"] = df["date_added"].str.strip()
# df.date_added
print(df.date_added.dtypes)

df["date_added"] = pd.to_datetime(df["date_added"])
```
