# PYHTON JOB LISTING SCRAPER

```python
# importing required libraries
import requests
from bs4 import BeautifulSoup
import csv
```

```python
# fetching the target url
url = "https://realpython.github.io/fake-jobs/"
response = requests.get(url)
response
```

```python
# we have successfully fetched the url
```

```python
# parse the HTML content using beautifulsoup
soup = BeautifulSoup(response.text, "html.parser")
print(soup)
```

```python
# finding all job listings
job_lists = soup.find_all("div", class_="card-content")
print(len(job_lists))
```

```python
job_lists
```

```python
# as we can see all the job titles are present inside the <h> tag
for i in job_lists:
    title = i.find("h2", class_="title is-5").text.strip()
    company = i.find("h3", class_="subtitle is-6 company").text.strip()
    location = i.find("p", class_="location").text.strip()

    # Extact the Apply link from the <a> tag inside the footer
    # we look for the href of the second link
    link = i.find_all("a")[1]["href"]

    print(f"job:{title}")
    print(f"Company: {company}")
    print(f"Location: {location}")
    print(f"Apply link: {link}")
```

```python
# getting the csv
import csv

with open("job_listings.csv", mode="w", newline="", encoding="utf-8") as file:
    job_csv = csv.writer(file)

    job_csv.writerow(["Job Title", "Company", "Location", "Apply Link"])
    
    for i in job_lists:
        title = i.find("h2", class_="title is-5").text.strip()
        company = i.find("h3", class_="subtitle is-6 company").text.strip()
        location = i.find("p", class_="location").text.strip()
        link = i.find_all("a")[1]["href"]
        
        # Write a row for each job
        job_csv.writerow([title, company, location, link])
```
